```python
import pandas as pd
df = pd.read_csv("../data/mobile_game_inapp_purchases.csv")
```

```python
# Data Quality Audit
df["UserID"].duplicated().sum() 
# No duplicate UserIDs, 3,024 players in the dataset.
```

#### 60 values missing in the player demographics, platform information columns.
#### 136 values missing in the purchase information columns.

```python
# Does a player with missing purchase information also have missing demographic information?
missing_purchase_information = df[df["InAppPurchaseAmount"].isnull()][["UserID", "Age", "Gender", "Country", "Device", "GameGenre", "SessionCount", "SpendingSegment", "InAppPurchaseAmount", "FirstPurchaseDaysAfterInstall", "PaymentMethod", "LastPurchaseDate"]].head(10)

missing_purchase_information
```

```python
df[df["InAppPurchaseAmount"].isnull()]["SpendingSegment"].value_counts(dropna=False)
```

#### **136 players don't have purchase information.** 
#### Since they don't have any records on the purchase, but they are still segmented, meaning that we just don't have their purchase behavior data, yet, they should not be assumed to have zero-spending.
#### With this assumption, we can either: (1) impute the missing values to better do the purchase-behavior analysis, or (2) exclude them from the analyses.
#### Only 136 out of more than 3k units, which is only a small percentage. Therefore, to avoid complications in our analysis, we can omit these players from the analysis of purchase behavior.

```python
df[df["Age"].isnull()][
    ["UserID", "Age", "Gender", "Country", "Device", "GameGenre"]
].head(10)
```

```python
df["Age"].isnull().sum()
```

#### Needed to check if 60 missing values in the demographic information are information of the same 60 players. 
#### Turned out it's a No. 
#### Therefore, keep these columns with missing values intact. Only when doing analysis in separate piece of information, we can drop the missing values in that column. 

```python
df.info()
```

```python
df.describe(include='all').T
```

```python
print("IAP = 0:", (df["InAppPurchaseAmount"] == 0).sum())

print("\nSpendingSegment among IAP = 0:")
print(df[df["InAppPurchaseAmount"] == 0]["SpendingSegment"].value_counts(dropna=False))

print("\nInvalid Age:")
print(((df["Age"] < 13) | (df["Age"] > 100)).sum())

print("\nInvalid SessionCount:")
print((df["SessionCount"] <= 0).sum())

print("\nInvalid AverageSessionLength:")
print((df["AverageSessionLength"] <= 0).sum())

print("\nInvalid FirstPurchaseDaysAfterInstall:")
print((df["FirstPurchaseDaysAfterInstall"] < 0).sum())
```

#### There are no invalid values in all columns.
#### What's left is to change the dtype of the "LastPurchaseDate" column.

```python
parsed_dates = pd.to_datetime(df["LastPurchaseDate"], errors='coerce')
print("\nInvalid LastPurchaseDate:", parsed_dates.isna().sum())
print("Min date:", parsed_dates.min())
print("Max date:", parsed_dates.max())
```

```python
df["LastPurchaseDate"] = pd.to_datetime(df["LastPurchaseDate"])
df["LastPurchaseDate"].dtype
df.info()
```

```python
# Check the value counts of categorical columns
categorical_cols = [
    "Gender",
    "Country",
    "Device",
    "GameGenre",
    "SpendingSegment",
    "PaymentMethod"
]
for col in categorical_cols:
    print(f"\n{col} value counts:")
    print(df[col].value_counts(dropna=False))
```
